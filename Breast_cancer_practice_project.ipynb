##### a plot of the counts againsts the gene ids

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt

top = expressed.sort_values("count", ascending=False).head(15)
sns.barplot(data=top, x="count", y="Geneid")
plt.show()
```

```python
plot = expressed.sort_values("sample.bam",ascending=False).head(15)
sns.barplot(data=plot, x="sample.bam", y="Geneid")
plt.show()
```
